#### logistic regression and regularisation

##### Load the breast cancer dataset from scikit-learn


```python
# import libraries
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
# Load the dataset 
cancer = load_breast_cancer()
```

```python
# View the available keys in the dataset
print("Keys:", cancer.keys())
```

```python
# Access data and targets
X = cancer.data    # Features (569 samples, 30 features)
y = cancer.target  # Labels (0 for malignant, 1 for benign)
```

```python
# Split the dataset into 80% training and 20% testing with random_state = 9001
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=9001
)

# Scale the data using MinMaxScaler
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Attempt logistic regression using the default solver
model = LogisticRegression()
model.fit(X_train_scaled, y_train)

# Predict on the test set and generate the classification report
y_pred = model.predict(X_test_scaled)
report = classification_report(y_test, y_pred)

print("Classification Report:")
print(report)
```
